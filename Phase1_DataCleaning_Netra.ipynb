```python
# ============================================================
# E-COMMERCE DATA ANALYTICS PROJECT
# CLEANING: ORDERS, RETURNS, SUPPLIERS & INVENTORY
# ============================================================

import pandas as pd
import numpy as np
import os

# ------------------------------------------------------------
# 1. LOAD REMAINING DATASETS
# ------------------------------------------------------------

orders = pd.read_csv("orders DA.csv")
returns = pd.read_csv("returns DA.csv")
suppliers = pd.read_csv("suppliers DA.csv")
inventory = pd.read_csv("inventory DA.csv")

print("Datasets loaded successfully.")
print("Orders:", orders.shape)
print("Returns:", returns.shape)
print("Suppliers:", suppliers.shape)
print("Inventory:", inventory.shape)


# ============================================================
# 2. ORDERS CLEANING
# ============================================================

print("\n========== ORDERS CLEANING ==========")

# Basic structure
print("\nShape:")
print(orders.shape)

print("\nMissing values:")
print(orders.isnull().sum())

print("\nDuplicate rows:")
print(orders.duplicated().sum())


# Remove completely duplicated rows only
orders = orders.drop_duplicates().copy()


# ------------------------------------------------------------
# ID VALIDATION
# ------------------------------------------------------------

print("\nDuplicate order_id:")
print(orders["order_id"].duplicated().sum())

# Check order ID format
invalid_order_id = ~orders["order_id"].astype(str).str.match(
    r"^ORD\d{6}$",
    na=False
)

print("\nInvalid order_id:")
print(invalid_order_id.sum())


# ------------------------------------------------------------
# DATE CLEANING
# ------------------------------------------------------------

date_columns = [
    "order_date",
    "delivered_date"
]

for col in date_columns:
    if col in orders.columns:
        orders[col] = pd.to_datetime(
            orders[col],
            format="%d-%m-%Y",
            errors="coerce"
        )

print("\nDate columns converted.")


# ------------------------------------------------------------
# NUMERIC COLUMN CLEANING
# ------------------------------------------------------------

numeric_columns = [
    "quantity",
    "unit_price",
    "gst_rate",
    "gst_amount",
    "discount_amount",
    "total_amount",
    "final_amount"
]

for col in numeric_columns:
    if col in orders.columns:
        orders[col] = pd.to_numeric(
            orders[col],
            errors="coerce"
        )


# ------------------------------------------------------------
# QUANTITY CHECK
# ------------------------------------------------------------

if "quantity" in orders.columns:

    print("\nInvalid quantity:")
    print((orders["quantity"] <= 0).sum())

    # Convert whole-number quantities to integer
    if orders["quantity"].notna().all():
        orders["quantity"] = orders["quantity"].astype(int)


# ------------------------------------------------------------
# PRICE / AMOUNT CHECKS
# ------------------------------------------------------------

for col in [
    "unit_price",
    "gst_amount",
    "discount_amount",
    "total_amount",
    "final_amount"
]:

    if col in orders.columns:

        print(f"\nNegative values in {col}:")
        print((orders[col] < 0).sum())


# ------------------------------------------------------------
# ORDER STATUS CHECK
# ------------------------------------------------------------

if "order_status" in orders.columns:

    print("\nOrder status:")
    print(orders["order_status"].value_counts(dropna=False))


# ------------------------------------------------------------
# DELIVERY DATE LOGIC
# delivered_date should not be before order_date
# ------------------------------------------------------------

if "order_date" in orders.columns and "delivered_date" in orders.columns:

    invalid_delivery_dates = (
        orders["delivered_date"].notna()
        & orders["order_date"].notna()
        & (orders["delivered_date"] < orders["order_date"])
    )

    print("\nDelivered date before order date:")
    print(invalid_delivery_dates.sum())


# ------------------------------------------------------------
# FINAL AMOUNT OUTLIER CHECK
# ------------------------------------------------------------

if "final_amount" in orders.columns:

    print("\nOrders above ₹5,00,000:")
    print((orders["final_amount"] > 500000).sum())


# ------------------------------------------------------------
# ORDERS FINAL VALIDATION
# ------------------------------------------------------------

print("\nFinal Orders missing values:")
print(orders.isnull().sum())

print("\nFinal Orders duplicates:")
print(orders.duplicated().sum())


# SAVE ORDERS
orders.to_csv("orders_cleaned.csv", index=False)

print("\nOrders cleaned and saved successfully.")


# ============================================================
# 3. RETURNS CLEANING
# ============================================================

print("\n========== RETURNS CLEANING ==========")

print("\nShape:")
print(returns.shape)

print("\nMissing values:")
print(returns.isnull().sum())

print("\nDuplicate rows:")
print(returns.duplicated().sum())

# Remove only completely duplicated rows
returns = returns.drop_duplicates().copy()


# ------------------------------------------------------------
# ID CHECKS
# ------------------------------------------------------------

if "return_id" in returns.columns:

    print("\nDuplicate return_id:")
    print(returns["return_id"].duplicated().sum())

    invalid_return_id = ~returns["return_id"].astype(str).str.match(
        r"^RET\d{6}$",
        na=False
    )

    print("\nInvalid return_id:")
    print(invalid_return_id.sum())


if "order_id" in returns.columns:

    invalid_return_order_id = ~returns["order_id"].astype(str).str.match(
        r"^ORD\d{6}$",
        na=False
    )

    print("\nInvalid return order_id:")
    print(invalid_return_order_id.sum())


# ------------------------------------------------------------
# DATE COLUMNS
# ------------------------------------------------------------

return_date_columns = [
    "return_date"
]

for col in return_date_columns:
    if col in returns.columns:
        returns[col] = pd.to_datetime(
            returns[col],
            format="%d-%m-%Y",
            errors="coerce"
        )


# ------------------------------------------------------------
# NUMERIC COLUMNS
# ------------------------------------------------------------

return_numeric_columns = [
    "refund_amount",
    "return_amount"
]

for col in return_numeric_columns:
    if col in returns.columns:
        returns[col] = pd.to_numeric(
            returns[col],
            errors="coerce"
        )

        print(f"\nNegative values in {col}:")
        print((returns[col] < 0).sum())


# ------------------------------------------------------------
# RETURN STATUS / REASON
# ------------------------------------------------------------

for col in ["return_status", "return_reason"]:

    if col in returns.columns:

        print(f"\n{col}:")
        print(returns[col].value_counts(dropna=False))


# ------------------------------------------------------------
# FINAL RETURNS VALIDATION
# ------------------------------------------------------------

print("\nFinal Returns missing values:")
print(returns.isnull().sum())

print("\nFinal Returns duplicates:")
print(returns.duplicated().sum())


# SAVE RETURNS
returns.to_csv("returns_cleaned.csv", index=False)

print("\nReturns cleaned and saved successfully.")


# ============================================================
# 4. SUPPLIERS CLEANING
# ============================================================

print("\n========== SUPPLIERS CLEANING ==========")

print("\nShape:")
print(suppliers.shape)

print("\nMissing values:")
print(suppliers.isnull().sum())

print("\nDuplicate rows:")
print(suppliers.duplicated().sum())

suppliers = suppliers.drop_duplicates().copy()


# ------------------------------------------------------------
# SUPPLIER ID
# ------------------------------------------------------------

if "supplier_id" in suppliers.columns:

    print("\nDuplicate supplier_id:")
    print(suppliers["supplier_id"].duplicated().sum())

    invalid_supplier_id = ~suppliers["supplier_id"].astype(str).str.match(
        r"^SUP\d{4}$",
        na=False
    )

    print("\nInvalid supplier_id:")
    print(invalid_supplier_id.sum())


# ------------------------------------------------------------
# NUMERIC COLUMNS
# ------------------------------------------------------------

supplier_numeric_columns = [
    "rating",
    "total_products",
    "total_orders"
]

for col in supplier_numeric_columns:

    if col in suppliers.columns:

        suppliers[col] = pd.to_numeric(
            suppliers[col],
            errors="coerce"
        )

        print(f"\nNegative values in {col}:")
        print((suppliers[col] < 0).sum())


# ------------------------------------------------------------
# EMAIL CHECK
# ------------------------------------------------------------

if "email" in suppliers.columns:

    invalid_email = ~suppliers["email"].astype(str).str.match(
        r"^[^@\s]+@[^@\s]+\.[^@\s]+$",
        na=False
    )

    print("\nInvalid supplier emails:")
    print(invalid_email.sum())


# ------------------------------------------------------------
# FINAL SUPPLIER VALIDATION
# ------------------------------------------------------------

print("\nFinal Suppliers missing values:")
print(suppliers.isnull().sum())

print("\nFinal Suppliers duplicates:")
print(suppliers.duplicated().sum())


# SAVE SUPPLIERS
suppliers.to_csv("suppliers_cleaned.csv", index=False)

print("\nSuppliers cleaned and saved successfully.")


# ============================================================
# 5. INVENTORY CLEANING
# ============================================================

print("\n========== INVENTORY CLEANING ==========")

print("\nShape:")
print(inventory.shape)

print("\nMissing values:")
print(inventory.isnull().sum())

print("\nDuplicate rows:")
print(inventory.duplicated().sum())

inventory = inventory.drop_duplicates().copy()


# ------------------------------------------------------------
# INVENTORY ID
# ------------------------------------------------------------

if "inventory_id" in inventory.columns:

    print("\nDuplicate inventory_id:")
    print(inventory["inventory_id"].duplicated().sum())


# ------------------------------------------------------------
# PRODUCT ID
# ------------------------------------------------------------

if "product_id" in inventory.columns:

    invalid_inventory_product_id = ~inventory["product_id"].astype(str).str.match(
        r"^PRD\d{4}$",
        na=False
    )

    print("\nInvalid product_id:")
    print(invalid_inventory_product_id.sum())


# ------------------------------------------------------------
# NUMERIC COLUMNS
# ------------------------------------------------------------

inventory_numeric_columns = [
    "stock_quantity",
    "reorder_level",
    "unit_cost"
]

for col in inventory_numeric_columns:

    if col in inventory.columns:

        inventory[col] = pd.to_numeric(
            inventory[col],
            errors="coerce"
        )

        print(f"\nNegative values in {col}:")
        print((inventory[col] < 0).sum())


# ------------------------------------------------------------
# STOCK STATUS
# ------------------------------------------------------------

if "stock_status" in inventory.columns:

    print("\nStock status:")
    print(inventory["stock_status"].value_counts(dropna=False))


# ------------------------------------------------------------
# OUT OF STOCK COUNT
# ------------------------------------------------------------

if "stock_status" in inventory.columns:

    print("\nOut of Stock products:")
    print(
        (inventory["stock_status"].astype(str).str.strip().str.lower()
         == "out of stock").sum()
    )


# ------------------------------------------------------------
# FINAL INVENTORY VALIDATION
# ------------------------------------------------------------

print("\nFinal Inventory missing values:")
print(inventory.isnull().sum())

print("\nFinal Inventory duplicates:")
print(inventory.duplicated().sum())


# SAVE INVENTORY
inventory.to_csv("inventory_cleaned.csv", index=False)

print("\nInventory cleaned and saved successfully.")


# ============================================================
# 6. CHECK ALL CLEANED FILES
# ============================================================

print("\n========== CLEANED FILES ==========")

cleaned_files = [
    "customers_cleaned.csv",
    "products_cleaned.csv",
    "payments_cleaned.csv",
    "order_items_cleaned.csv",
    "orders_cleaned.csv",
    "returns_cleaned.csv",
    "suppliers_cleaned.csv",
    "inventory_cleaned.csv"
]

for file in cleaned_files:

    if os.path.exists(file):

        df = pd.read_csv(file)

        print(
            f"{file:<30} "
            f"Rows: {df.shape[0]:<8} "
            f"Columns: {df.shape[1]}"
        )

    else:

        print(f"{file:<30} NOT FOUND")


print("\n========== CLEANING COMPLETE ==========")
```
